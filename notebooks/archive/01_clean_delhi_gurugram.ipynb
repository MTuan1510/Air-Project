## Bước 2: Xử lý missing values

Cách làm:

1. **Săn missing ẩn** — kiểm tra các chuỗi giá trị `0` liên tiếp dài (≥3 ngày): nồng độ khí thực tế hiếm khi đứng yên ở đúng 0 nhiều ngày liền, nhiều khả năng là cảm biến bị treo (flatline), không phải giá trị thật → chuyển các chuỗi này thành `NaN`.
2. **Không coi mọi missing như nhau** — nhiều cột có khối thiếu **liên tục hàng chục–hàng nghìn ngày** (cảm biến chưa lắp/hỏng dài hạn), khác hẳn với thiếu rải rác vài ngày. Missing phụ thuộc vào `Date` (biến quan sát được) → cơ chế **MAR**, không phải MCAR.
3. **Chỉ nội suy khoảng trống ngắn** (mặc định ≤14 ngày) — khoảng trống dài **CHỦ Ý để nguyên `NaN`**, không bịa số cho hàng trăm/nghìn ngày không có dữ liệu thật.

```python
# Chẩn đoán: cột nào nên nội suy trực tiếp, cột nào có khối lớn cần cẩn thận, cột nào nên đề xuất DROP
print("=== DELHI - diagnose_missing ===")
display(diagnose_missing(delhi))

print("\n=== GURUGRAM - diagnose_missing ===")
display(diagnose_missing(gurugram))
```

**Đọc bảng trên:** `NH3` và `Xylene` ở Gurugram đều ~93% missing và gần như toàn bộ nằm trong 1 khối duy nhất (block_ratio ~100%). Ban đầu mình đề xuất DROP cả 2 cột cho Gurugram — nhưng đã kiểm tra lại số liệu thật của 6 thành phố còn lại trong file gốc (xem bảng dưới) và sửa lại quyết định:

| City | NH3 %missing | Xylene %missing |
|---|---|---|
| Delhi | 0.4% | 38.9% |
| Gurugram | 92.9% | 92.9% |
| Lucknow | 50.0% | 100.0% |
| Jaipur | 1.3% | 100.0% |
| Bengaluru | 10.1% | 100.0% |
| Amritsar | 3.4% | 17.4% |
| Chennai | 23.0% | 100.0% |
| Hyderabad | 17.9% | 6.3% |

- **NH3**: đa số thành phố có dữ liệu khá tốt (0.4%–23%), chỉ Gurugram và Lucknow là tệ → **giữ lại cột này cho cả 8 thành phố**, drop sẽ làm mất dữ liệu tốt của 6 thành phố kia.
- **Xylene**: **4/8 thành phố thiếu 100%**, cả Delhi lẫn Gurugram đều thiếu nặng (38.9%, 92.9%) → **đề xuất DROP cột này cho TẤT CẢ 8 thành phố**, chỉ Amritsar/Hyderabad có dữ liệu tạm ổn nhưng không đủ để giữ cột cho cả bộ dữ liệu chung.

`PM10` ở Gurugram còn 55.9% với khối lớn chiếm 81% — vẫn giữ lại (không đủ điều kiện DROP theo ngưỡng 90%) nhưng cần lưu ý rõ trong báo cáo là dữ liệu không đầy đủ.

⚠️ **Cần chốt với nhóm:** đề xuất DROP `Xylene` cho cả 8 thành phố (không riêng Gurugram), giữ `NH3`. Và ngưỡng `max_short_gap=14` ngày cho nội suy.

```python
delhi_clean, delhi_flat_changed = impute_time_series(delhi, "Delhi", drop_cols=["Xylene"])
print()
gurugram_clean, gurugram_flat_changed = impute_time_series(
    gurugram, "Gurugram", drop_cols=["Xylene"]  # ĐỀ XUẤT - đã đồng ý với nhóm chưa?
)
```

```python
# Kiểm chứng: so sánh % missing trước / sau cho từng cột
pollutant_cols = ['PM2.5','PM10','NO','NO2','NOx','NH3','CO','SO2','O3','Benzene','Toluene','Xylene']

def compare_missing(before_df, after_df, cols):
    before = before_df[cols].isna().mean() * 100
    after_cols = [c for c in cols if c in after_df.columns]
    after = after_df[after_cols].isna().mean() * 100
    return pd.DataFrame({'truoc_%': before.round(1), 'sau_%': after.reindex(before.index).round(1)})

print("=== DELHI: truoc/sau ===")
display(compare_missing(delhi, delhi_clean, pollutant_cols + ['AQI']))

print("\n=== GURUGRAM: truoc/sau ===")
display(compare_missing(gurugram, gurugram_clean, pollutant_cols + ['AQI']))
```

**Lưu ý:** một vài cột (CO, Toluene ở Delhi; Benzene, Toluene ở Gurugram) có `sau_%` **cao hơn** `truoc_%` — không phải lỗi, mà vì bước phát hiện flatline vừa lộ ra các đoạn `0` giả trước đây bị tính nhầm là "có dữ liệu". Đây là phát hiện thật, cần ghi vào cleaning log để báo cáo với nhóm/giảng viên, không nên ỉm đi.

```python
os.makedirs("../data/interim", exist_ok=True)
delhi_clean.to_csv("../data/interim/delhi_after_missing.csv", index=False)
gurugram_clean.to_csv("../data/interim/gurugram_after_missing.csv", index=False)
print("Đã lưu data/interim/delhi_after_missing.csv và data/interim/gurugram_after_missing.csv")
print("Delhi:", delhi_clean.shape, " | Gurugram:", gurugram_clean.shape)
```

## Bước 3: Outliers & validity checks (W4)

Quyết định đã chọn cho bước này (an toàn, chưa xoá gì):
- Phát hiện bằng **IQR**, tính ngưỡng **riêng theo từng tháng dương lịch** (gộp các năm) — tránh gắn nhầm mùa ô nhiễm cao (đốt rơm rạ tháng 10-11, Diwali) thành outlier.
- **Chỉ đánh dấu** (`<cột>_is_outlier`), **không xoá/sửa giá trị gốc** — quyết định giữ/xoá để dành cho lúc gộp đủ 8 thành phố, đối chiếu xem outlier là lỗi cảm biến hay sự kiện thật.
- Không áp dụng cho `AQI` — đây là biến kết quả, các giá trị AQI cao chính là điều nhóm muốn phân tích, không phải lỗi.
- Kèm theo bộ **validity checks** cơ bản (không âm, ngày không ở tương lai, AQI khớp AQI_Bucket) để kiểm tra nhanh sau mỗi lần xử lý.

```python
from utils import flag_outliers_iqr, validity_checks

delhi_outliers = flag_outliers_iqr(delhi_clean)
gurugram_outliers = flag_outliers_iqr(gurugram_clean)

for name, df in [("DELHI", delhi_outliers), ("GURUGRAM", gurugram_outliers)]:
    outlier_cols = [c for c in df.columns if c.endswith('_is_outlier')]
    print(f"=== {name}: số ngày bị đánh dấu outlier mỗi cột ===")
    display(df[outlier_cols].sum().sort_values(ascending=False))
```

```python
print("=== DELHI: validity checks ===")
display(validity_checks(delhi_outliers))

print("\n=== GURUGRAM: validity checks ===")
display(validity_checks(gurugram_outliers))
```
